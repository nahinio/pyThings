### Unpacking a List

```python
person = ['Batman', 26, 'Software Engineer', 'Spain']
#name = person[0]
#age = person[1]
#role = person[2]
#country = person[3]

name, age, role, country = person

print(name)
print(age)
print(role)
print(country)
```

```python
person = ['Batman', 26, 'Software Engineer', 'Spain']

# Using Asterisk to control the Unpacking
name, *unnecessary, country = person    # Only one * is allowed in unpacking

print(name)
print(unnecessary)
print(country)

```

```python
person = ['Batman']

name, *unnecessary = person

print(name)
print(unnecessary)
```

```python
person = 'Batman'

name, *unnecessary = person

print(name)
print(unnecessary)
```

Unpacking Using '_' underscore

```python
person = ['Batman', 26, 'Software Engineer', 'Spain']

name, _, role, _ = person    # Can use multiple underscore '_'

print(name)
print(role)
```

### Add, Remove, and Update Lists in Python

Add Items

```python
list = ['a', 'b', 'c']

list.append('d')
list.append('e')
print(list)


list.insert(3, 'x')
print(list)
```

```python
matrix = [['a', 'b', 'c'],
          ['d', 'e', 'f'],
          ['g', 'h', 'i']]

matrix.append(['j', 'k', 'l'])    # Add new row
print(matrix)

matrix.insert(1, ['q', 'm', 'b'])
print(matrix)

matrix[1].append('o')
print(matrix)

matrix[1].insert(1, 'p')
print(matrix)
```

Remove Items

```python
list = ['a', 'b', 'c', 'd', 'e']

print(list)
list.clear()
print(list)
```

```python
list = ['a', 'b', 'c', 'd', 'e']

# Remove by value
list.remove('c')    # Remove the first match
print(list)

# Remove by Position
list.pop()    # Default remove last item
list.pop(0)   # Remove from exact index
print(list)
```

```python
matrix = [['a', 'b', 'c'],
          ['d', 'e', 'f'],
          ['g', 'h', 'i']]

matrix.remove(['d', 'e', 'f'])    # Remove row by value
print(matrix)

matrix.pop()    # Remove last row
print(matrix)

matrixNew = [['a', 'b', 'c'],
          ['d', 'e', 'f'],
          ['g', 'h', 'i']]

# Remove specific value
matrixNew[1].remove('e')
print(matrixNew)

# Remove specific positioned value
matrixNew[-1].pop(0)
print(matrixNew)

matrixNew[0].pop()
print(matrixNew)
```

### Order Lists in Python

```python
list = ['q', 'w', 'e', 'r', 't']

list.sort()
print(list)

list.sort(reverse=True)
print(list)
```

```python
matrix = [['q', 'a', 'c'],
          ['d', 'e', 'f'],
          ['a', 'a', 'i']]

matrix.sort()    # Sorts by the first item of each row (Basically sort column )
print(matrix)

matrixNew = [['q', 'a', 'c'],
          ['d', 'e', 'f'],
          ['a', 'a', 'i']]
matrixNew[0].sort()    # Sorts by specific row
print(matrixNew)
```

```python
list = ['q', 'w', 'e', 'r', 't']

listSorted = sorted(list)    # Returns a new sorted list
print(listSorted)
print(list)    # Original list remains unchanged
```

```python
list = ['q', 'w', 'e', 'r', 't']

list.reverse()    # Reverses the list in place
print(list)

list = ['q', 'w', 'e', 'r', 't']

listReversed = reversed(list)    # Returns a new reversed list
print(listReversed)
print(list)    # Original list remains unchanged
```

### Copy Python Lists

```python
list = ['q', 'w', 'e', 'r', 't']
newList = list    # It does not create a new list, it just creates a new reference to the same list in memory. So any changes made to newList will affect list

print(newList)

list.append('y')
newList.append('z')

print(list)
print(newList)
```

```python
list = ['q', 'w', 'e', 'r', 't']
listCopy = list.copy()    # Creates a new copy of the list in memory. So any changes made to listCopy will not affect list

listCopy.append('y')
print(list)
print(listCopy)
```

```python
matrix = [['q', 'c'],
          ['d', 'e']
]

matrixCopy = matrix.copy()

matrixCopy[0].append('y')    # Changes made to the inner list of matrixCopy will also affect the inner list of matrix, because the inner lists are still references to the same objects in memory. This is known as a shallow copy.

print(matrix)
print(matrixCopy)

import copy

newMatrix = [['q', 'c'],
          ['d', 'e']
]

newMatrixCopy = copy.deepcopy(newMatrix)
newMatrixCopy[0].append('y')    # Changes made to the inner list of newMatrixCopy will not affect the inner list of newMatrix, because copy.deepcopy creates a new object in memory. This is known as a deep copy.

print(newMatrix)
print(newMatrixCopy)

```

is Operator

```python
import copy
original = [['q', 'c'],
          ['d', 'e']
] 

# Assignment
copy1 = original
print(copy1 is original)    # True, because copy1 is just a reference to the same list in memory as original
print(copy1[0] is original[0])    # True, because copy1[0] is just a reference to the same inner list in memory as original[0]

# Shallow Copy
copy2 = original.copy()
print(copy2 is original)    # False, because copy2 is a new list in memory, but the inner lists are still references to the same objects in memory. This is known as a shallow copy.
print(copy2[0] is original[0])    # True, because copy2[0] is just a reference to the same inner list in memory as original[0]

# Deep Copy
import copy
copy3 = copy.deepcopy(original)
print(copy3 is original)    # False, because copy3 is a new list in memory, and the inner lists are also new objects in memory. This is known as a deep copy.
print(copy3[0] is original[0])    # False, because copy3[0] is a new object in memory, and is not a reference to the same inner list as original[0]
```

### Combine Lists

```python
number = [1, 2, 3, 4, 5]
letters = ['a', 'b', 'c', 'd', 'e']

numbers_and_letters = number + letters
print(numbers_and_letters)

multiple_copies = number * 3    # Creates a new list with 3 copies of the original list in memory. 
print(multiple_copies)

# Lists inside a list
nested_lists = [number, letters]
print(nested_lists)

# Extending a list with another list
number.extend(letters)    # type: ignore
print(number) 


# List of tuples
number_again = [1, 2, 3, 4, 5]
letters_again = ['a', 'b', 'c', 'd', 'e']

comb = zip(number_again, letters_again)
print(comb)    # <zip object at 0x000001F3D8C1A4C0>
print(list(comb))    # type: ignore


comb_again = zip(number_again, letters_again, "man")
print(list(comb_again))    # type: ignore 
```
