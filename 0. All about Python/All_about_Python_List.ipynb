### Indexing and Slicing

```python
yo_list = ['a', 'b', 'c', 'd']

# First Item
print(yo_list[0])

# Last Item
print(yo_list[-1])

# Slicing
print(yo_list[1:3])

# First 2 Items
print(yo_list[:2])

# Last 2 Items
print(yo_list[-2:])
```

```python
matrix = [['a', 'b', 'c'],
          ['d', 'e', 'f'],
          ['g', 'h', 'i']]

print(matrix[0])    #Access row
print(matrix[0][1])    #Random item
print(matrix)    #Full matrix
print(matrix[-1])    #Last row
print(matrix[-1][-1])    #Last row
```

### Unpacking a List

```python
person = ['Batman', 26, 'Software Engineer', 'Spain']
#name = person[0]
#age = person[1]
#role = person[2]
#country = person[3]

name, age, role, country = person

print(name)
print(age)
print(role)
print(country)
```

```python
person = ['Batman', 26, 'Software Engineer', 'Spain']

# Using Asterisk to control the Unpacking
name, *unnecessary, country = person    # Only one * is allowed in unpacking

print(name)
print(unnecessary)
print(country)

```

```python
person = ['Batman']

name, *unnecessary = person

print(name)
print(unnecessary)
```

```python
person = 'Batman'

name, *unnecessary = person

print(name)
print(unnecessary)
```

Unpacking Using '_' underscore

```python
person = ['Batman', 26, 'Software Engineer', 'Spain']

name, _, role, _ = person    # Can use multiple underscore '_'

print(name)
print(role)
```

```python
person = ['Batman', 26, 'Software Engineer', 'CS Undergrad', 'Spain']

# Ignoring the middle (*_)
name, *_, country = person

print(name)
```

### Expliore and Analyze List

```python
numbers = [1, 2, 6, 4, 3, 4]

print(len(numbers))
print(min(numbers))
print(max(numbers))
print(sum(numbers))

print(all(numbers))
print(all([1, 0, 3]))
print(all(['a', '', 'c']))

print(any(numbers))
print(any([1, 0, 3]))
print(any(['a', '', 'c']))
print(all([0, 0, 0]))

print(numbers.count(4))
print(numbers.index(4))    # First appearence of the value

print(4 in numbers)
print(5 in numbers)

list1 = [1, 2, 3]
list2 = [1, 2, 3]

print(list1 == list2)

list3 = [5, 1, 1]
print(list1 < list3)    # First value only

print(list1 is list2)    # Different memory location
```

### Add, Remove, and Update Lists in Python

Add Items

```python
list = ['a', 'b', 'c']

list.append('d')
list.append('e')
print(list)


list.insert(3, 'x')
print(list)
```

```python
matrix = [['a', 'b', 'c'],
          ['d', 'e', 'f'],
          ['g', 'h', 'i']]

matrix.append(['j', 'k', 'l'])    # Add new row
print(matrix)

matrix.insert(1, ['q', 'm', 'b'])
print(matrix)

matrix[1].append('o')
print(matrix)

matrix[1].insert(1, 'p')
print(matrix)
```

Remove Items

```python
list = ['a', 'b', 'c', 'd', 'e']

print(list)
list.clear()
print(list)
```

```python
list = ['a', 'b', 'c', 'd', 'e']

# Remove by value
list.remove('c')    # Remove the first match
print(list)

# Remove by Position
list.pop()    # Default remove last item
list.pop(0)   # Remove from exact index
print(list)
```

```python
matrix = [['a', 'b', 'c'],
          ['d', 'e', 'f'],
          ['g', 'h', 'i']]

matrix.remove(['d', 'e', 'f'])    # Remove row by value
print(matrix)

matrix.pop()    # Remove last row
print(matrix)

matrixNew = [['a', 'b', 'c'],
          ['d', 'e', 'f'],
          ['g', 'h', 'i']]

# Remove specific value
matrixNew[1].remove('e')
print(matrixNew)

# Remove specific positioned value
matrixNew[-1].pop(0)
print(matrixNew)

matrixNew[0].pop()
print(matrixNew)
```
